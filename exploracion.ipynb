# Exploracion de datos - Adult / Census Income

EDA breve del dataset usado para entrenar el modelo.

```
import pandas as pd
from sklearn.datasets import fetch_openml

ds = fetch_openml("adult", version=2, as_frame=True)
df = ds.frame.copy()
df.columns = [x.replace("-", "_").replace(".", "_") for x in df.columns]
print(df.shape)
df.head()
```

## Balance de la variable objetivo

```
df["class"].value_counts(normalize=True)
```

## Estadisticos de variables numericas

```
df[["age","hours_per_week","capital_gain","capital_loss"]].describe()
```

## Frecuencias por categoria y nulos

```
print(df["workclass"].value_counts(dropna=False).head())
print()
print("Nulos por columna:")
print(df.isna().sum())
```
